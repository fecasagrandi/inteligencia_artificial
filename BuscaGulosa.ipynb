```python
class vertice:
    def __init__(self, rotulo, distancia_objetivo):
        self.rotulo = rotulo
        self.visitado = False
        self.distancia_objetivo = distancia_objetivo
        self.adjacente = []

    def adiciona_adjacente(self, adjacente):
        self.adjacente.append(adjacente)

    def mostra_adjacente(self):
        for adj in self.adjacente:
            print(adj.vertice.rotulo, adj.custo)
```

```python
class Adjacente:
    def __init__(self, vertice, custo):
        self.vertice = vertice
        self.custo = custo
```

```python
import numpy as np


class VetorOrdenado:
    def __init__(self, capacidade: int):
        self.capacidade = capacidade
        self.ultima_posicao = -1
        self.valores = np.empty(self.capacidade, dtype=object)

    def imprime(self):
        if self.ultima_posicao == -1:
            print("Vetor vazio")
            return
        else:       
            for i in range(self.ultima_posicao + 1):
                print(i, " - ", self.valores[i].rotulo, ' - ', self.valores[i].distancia_objetivo)
        

    def insere(self, vertice):
        if self.ultima_posicao == self.capacidade - 1:
            print("Capacidade máxima atingida")
            return

        if self.ultima_posicao == -1:
            posicao = 0

        for i in range(self.ultima_posicao + 1):
            posicao = i
            if self.valores[i].distancia_objetivo > vertice.distancia_objetivo:
                break

            if i == self.ultima_posicao:
                posicao = i + 1

        for i in range(self.ultima_posicao, posicao - 1, -1):
            self.valores[i + 1] = self.valores[i]

        self.valores[posicao] = vertice
        self.ultima_posicao += 1
```

```python
vetor = VetorOrdenado(5)
vetor.insere(grafo.arad)
vetor.insere(grafo.craiova)
vetor.insere(grafo.bucharest)
vetor.insere(grafo.drobeta)

vetor.imprime()
```
